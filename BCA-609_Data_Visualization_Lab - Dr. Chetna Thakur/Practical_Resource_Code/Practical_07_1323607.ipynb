### Program No. 07.  >> Area Chart for Time-Series Data
**Objective**: To visualize cumulative data over time.

Tasks:
- Create an area chart for yearly profit or rainfall data

**Outcome**: Students will understand growth patterns using area charts.


Import libraries

```python
# Tool: Python (Matplotlib)
import pandas as pd
import matplotlib.pyplot as plt
```

**Load Dataset**: `MMDU_Student_Performance`

```python
# Load dataset
df = pd.read_csv("MMDU_Student_Performance.csv")
```

Select required columns

```python
# Select required columns
data = df[['academic_year', 'percentage']]
```

Handle Missing Values

```python
# Remove missing values
data = data.dropna()
```

For Ploting purpose convert year_Data to string

```python
# Convert academic_year to string (for plotting)
data['academic_year'] = data['academic_year'].astype(str)
```

Calculate average percentage per academic year

```python
# Calculate average percentage per academic year
yearly_avg = data.groupby('academic_year')['percentage'].mean().sort_index()
```

Create Area Chart

```python
# Area chart for average percentage per academic year
plt.figure(figsize=(10, 5))
x = yearly_avg.index.astype(str)
y = yearly_avg.values
plt.fill_between(x, y, alpha=0.4, color='tab:blue') #type: ignore
plt.plot(x, y, marker='o', color='tab:blue')
plt.xlabel('Academic Year')
plt.ylabel('Average Percentage')
plt.title('Average Student Performance Over Academic Years (Area Chart)')
plt.grid(alpha=0.3)
plt.ylim(0, max(y) * 1.1)
plt.tight_layout()
plt.show()
```
