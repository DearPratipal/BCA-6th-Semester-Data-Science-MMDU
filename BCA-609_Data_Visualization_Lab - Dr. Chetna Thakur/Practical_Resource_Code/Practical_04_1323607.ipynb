1. Identify visualization goals for a dataset
2. Create a donut chart to represent proportional data

Outcome: Students will understand how visualization supports decision-making.


```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
```

Dataset Used: `MMU_Student_Performance.csv`

Load dataset

```python
df = pd.read_csv("MMDU_Student_Performance.csv")
```

Select column for proportional analysis

```python
specialization_counts = df['specialization'].value_counts()
```

Create donut chart

```python
plt.figure(figsize=(7, 7))
plt.pie(
    specialization_counts,
    labels=specialization_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4)
)
plt.title("Distribution of Students by Specialization (Donut Chart)")
plt.show()
```
