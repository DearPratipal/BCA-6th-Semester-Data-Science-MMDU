**Program No. 15.  >> Multivariate Chart Visualization**

**Objective:** To visualize more than two variables simultaneously.

**Tasks:**
- Create a multivariate chart using marks in three subjects

**Outcome:** Students will analyze complex datasets visually and understand relationships among multiple variables.


**Step 1:** Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

**Step 2:** Load and Prepare Dataset

```python
df = pd.read_csv("student_extended.csv")

df.columns = df.columns.str.strip().str.lower()

# Convert required columns to numeric
df["percentage"] = pd.to_numeric(df["percentage"], errors="coerce")
df["sgpa"] = pd.to_numeric(df["sgpa"], errors="coerce")
df["attendance_percentage"] = pd.to_numeric(df["attendance_percentage"], errors="coerce")
df["study_hours_per_day"] = pd.to_numeric(df["study_hours_per_day"], errors="coerce")

# Remove missing values
df = df.dropna(subset=[
    "percentage",
    "sgpa",
    "attendance_percentage",
    "study_hours_per_day"
])
```

**Step 3:** Create Multivariate Bubble Chart

```python
plt.figure(figsize=(9,6))

scatter = plt.scatter(
    df["study_hours_per_day"],          # X-axis
    df["percentage"],                   # Y-axis
    s=df["sgpa"] * 40,                  # Bubble size (SGPA)
    c=df["attendance_percentage"],      # Color (Attendance)
    cmap="viridis",
    alpha=0.7,
    edgecolors="black"
)

plt.title("Multivariate Academic Performance Visualization",
          fontsize=14,
          fontweight="bold")

plt.xlabel("Study Hours per Day")
plt.ylabel("Percentage (%)")

cbar = plt.colorbar(scatter)
cbar.set_label("Attendance Percentage")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Student Names
students = ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona"]

# Marks in 3 Subjects
math_marks = [88, 72, 95, 60, 85, 78]
science_marks = [75, 80, 88, 65, 90, 82]
english_marks = [92, 68, 78, 85, 72, 90]
```

```python
x = np.arange(len(students))
width = 0.25

plt.figure(figsize=(9,6))

bar1 = plt.bar(x - width, math_marks, width, label="Math")
bar2 = plt.bar(x, science_marks, width, label="Science")
bar3 = plt.bar(x + width, english_marks, width, label="English")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks in 3 Subjects — Grouped Bar Chart",
          fontsize=14, fontweight="bold")

plt.xticks(x, students)
plt.ylim(0, 100)
plt.legend()

# Add value labels
for bars in [bar1, bar2, bar3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2,
                 height + 1,
                 str(height),
                 ha='center',
                 va='bottom',
                 fontsize=8)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

**From the visualization:**
- Students studying more hours generally achieve higher percentages.
- Larger bubbles indicate higher SGPA.
- Darker color indicates better attendance.
- High-performing students tend to cluster in the upper-right region.
- Students with low attendance often show lower academic performance.

**This multivariate chart helps in:**
- Identifying performance patterns
- Detecting outliers
- Understanding interaction between variables
- Analyzing academic behavior visually
