Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

**Uncleaned_Student_Dataset** : Selected Attributes: `department`, `attendance_percentage`

Create a simple dataset

```python

# sample data (replace with your real dataset if available)
df = pd.DataFrame({
    'department': ['Math', 'Physics', 'Chemistry', 'Biology', 'English'],
    'attendance_percentage': [88, 92, 79, 85,  ninety := 90]  # replace or load real values
})
```

Save for external tools

```python
# save for external tools
df.to_csv('Uncleaned_Student_Dataset_Copy.csv', index=False)
```

Create Visualization

```python
# quick in-notebook visualization (bar chart of average attendance per department)
agg = df.groupby('department', as_index=False)['attendance_percentage'].mean()
ax = agg.plot(kind='bar', x='department', y='attendance_percentage', legend=False, rot=45)
ax.set_ylabel('Attendance (%)')
ax.set_title('Average Attendance by Department')
plt.tight_layout()
plt.show()
```
