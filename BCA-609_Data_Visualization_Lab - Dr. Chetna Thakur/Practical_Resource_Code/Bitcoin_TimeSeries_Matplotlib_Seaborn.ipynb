# 📈 Bitcoin Time-Series Analysis (Matplotlib & Seaborn)
Professional financial visualization using matplotlib and seaborn.

Includes:
- Close Price Area Chart
- Candlestick-style Line Visualization
- Volume Trend
- Moving Average Analysis
- Combined Price & Volume Subplots


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14,6)
```

```python
# Load dataset
df = pd.read_csv('bitcoin_2017_to_2023.csv')

# Convert timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp')

df.head()
```

## 📊 Close Price Area Chart

```python
plt.figure()
plt.fill_between(df['timestamp'], df['close'], alpha=0.3)
plt.plot(df['timestamp'], df['close'], linewidth=1.5)
plt.title('Bitcoin Close Price Trend')
plt.xlabel('Time')
plt.ylabel('Close Price (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 📊 Volume Trend

```python
plt.figure()
sns.lineplot(x='timestamp', y='volume', data=df)
plt.title('Bitcoin Trading Volume Over Time')
plt.xlabel('Time')
plt.ylabel('Volume')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 📈 Moving Average (20 Period)

```python
df['MA_20'] = df['close'].rolling(window=20).mean()

plt.figure()
sns.lineplot(x='timestamp', y='close', data=df, label='Close Price')
sns.lineplot(x='timestamp', y='MA_20', data=df, label='20 Period MA')
plt.title('Bitcoin Close Price with Moving Average')
plt.xlabel('Time')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

## 📊 Combined Price & Volume Subplots

```python
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(14,8))

# Close Price
ax1.plot(df['timestamp'], df['close'])
ax1.set_title('Bitcoin Close Price')
ax1.set_ylabel('Price (USD)')

# Volume
ax2.bar(df['timestamp'], df['volume'])
ax2.set_title('Trading Volume')
ax2.set_ylabel('Volume')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
