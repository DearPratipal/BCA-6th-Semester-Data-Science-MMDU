**Objective:** To visualize hierarchical and categorical data.

**Tasks:**
- Design a tree map for department-wise student count

**Outcome:** Students will understand hierarchical data visualization.


**Step 1: Import Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify
```

**Step 2: Load Relevant Dataset**

```python
# Load dataset
df = pd.read_csv("student_extended.csv")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Check available columns
print(df.columns)
```

**Step 3: Prepare Hierarchical Data**

```python
# Make sure 'department' column exists
if "department" not in df.columns:
    print("Column 'department' not found. Available columns:")
    print(df.columns)
else:
    # Create department-wise student count
    dept_counts = df["department"].value_counts().reset_index()
    dept_counts.columns = ["Department", "Student_Count"]

    # Create Tree Map
    plt.figure(figsize=(10,6))

    sizes = dept_counts["Student_Count"]

    labels = [
        f"{dept}\n{count} Students"
        for dept, count in zip(dept_counts["Department"],
                               dept_counts["Student_Count"])
    ]

    colors = plt.cm.Paired(range(len(sizes)))

    squarify.plot(
        sizes=sizes,
        label=labels,
        color=colors,
        alpha=0.85
    )

    plt.title(
        "Department-wise Student Distribution — Tree Map",
        fontsize=14,
        fontweight="bold"
    )

    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
# Objective:
# Design a secured data visualization system architecture
# and explain the security measures used.

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ---------------------------------------------------------
# Create Canvas
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(16, 10))
ax.set_xlim(0, 16)
ax.set_ylim(0, 11)
ax.axis('off')

ax.set_title(
    'Secured Data Visualization System Architecture',
    fontsize=16,
    fontweight='bold',
    pad=20
)

# ---------------------------------------------------------
# Function to Draw Box
# ---------------------------------------------------------
def draw_box(ax, x, y, w, h, label, color, sublabel=''):
    rect = mpatches.FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.15",
        facecolor=color,
        edgecolor='black',
        linewidth=1.5
    )
    ax.add_patch(rect)

    ax.text(
        x + w/2, y + h/2 + 0.2,
        label,
        ha='center',
        va='center',
        fontweight='bold',
        fontsize=9
    )

    if sublabel:
        ax.text(
            x + w/2, y + h/2 - 0.3,
            sublabel,
            ha='center',
            va='center',
            fontsize=7,
            style='italic'
        )

# ---------------------------------------------------------
# Layer 1 – Presentation & Access Control
# ---------------------------------------------------------
draw_box(ax, 1, 9, 3, 1.2, 'Users / Clients', '#BBDEFB', 'Web / Mobile / API')
draw_box(ax, 5, 9, 3, 1.2, 'Authentication', '#C8E6C9', 'OAuth2 / JWT / MFA')
draw_box(ax, 9, 9, 3, 1.2, 'Authorization', '#FFF9C4', 'RBAC / ABAC')

# ---------------------------------------------------------
# Layer 2 – Application & Security
# ---------------------------------------------------------
draw_box(ax, 0.5, 6.5, 3, 1.2, 'API Gateway', '#E1BEE7', 'Rate Limiting / WAF')
draw_box(ax, 4, 6.5, 3.5, 1.2, 'Visualization Engine', '#B2EBF2', 'Plotly / D3.js')
draw_box(ax, 8.5, 6.5, 3.5, 1.2, 'Encryption Layer', '#FFCCBC', 'TLS 1.3 / AES-256')

# ---------------------------------------------------------
# Layer 3 – Data & Monitoring
# ---------------------------------------------------------
draw_box(ax, 0.5, 3.8, 3, 1.2, 'Database', '#D1C4E9', 'Encrypted at Rest')
draw_box(ax, 4, 3.8, 3.5, 1.2, 'Data Processing', '#B2DFDB', 'Masking / Anonymization')
draw_box(ax, 8.5, 3.8, 3.5, 1.2, 'Audit Logging', '#FFE0B2', 'Access Logs / SIEM')

# ---------------------------------------------------------
# Layer 4 – Infrastructure
# ---------------------------------------------------------
draw_box(ax, 2, 1.5, 4, 1.2, 'Cloud Infrastructure', '#F0F4C3', 'VPC / Firewall / IDS')
draw_box(ax, 7, 1.5, 5, 1.2, 'Backup & Disaster Recovery', '#FFCDD2', 'Encrypted / Geo-Redundant')

# ---------------------------------------------------------
# Arrows (Data Flow)
# ---------------------------------------------------------
arrows = [
    (2.5,9,2.5,7.7), (6.5,9,6.5,7.7), (10.5,9,10.5,7.7),
    (2,6.5,2,5), (5.75,6.5,5.75,5), (10.25,6.5,10.25,5),
    (2,3.8,4,2.7), (5.75,3.8,5.75,2.7), (10.25,3.8,9.5,2.7)
]

for x1, y1, x2, y2 in arrows:
    ax.annotate(
        '',
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(arrowstyle='->', color='gray', lw=1.5)
    )

# ---------------------------------------------------------
# Layer Labels (Right Side)
# ---------------------------------------------------------
layer_labels = [
    (13.5, 9.5, 'Layer 1:\nPresentation & Access Control'),
    (13.5, 7.0, 'Layer 2:\nApplication & Security'),
    (13.5, 4.3, 'Layer 3:\nData & Monitoring'),
    (13.5, 2.0, 'Layer 4:\nInfrastructure')
]

for x, y, label in layer_labels:
    ax.text(
        x, y, label,
        fontsize=8,
        ha='center',
        style='italic',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8)
    )

plt.tight_layout()
plt.show()

# =========================================================
# SECURITY MEASURES EXPLANATION
# =========================================================

print("\n" + "=" * 65)
print("SECURITY MEASURES IN THE DATA VISUALIZATION SYSTEM")
print("=" * 65)

measures = [
    ("1. Authentication & Authorization",
     "OAuth2/JWT tokens with Multi-Factor Authentication (MFA)\n"
     "Role-Based Access Control (RBAC) restricts user privileges"),

    ("2. Data Encryption",
     "TLS 1.3 encrypts data in transit\n"
     "AES-256 encryption protects data at rest"),

    ("3. API Security",
     "API Gateway provides rate limiting and request validation\n"
     "Web Application Firewall (WAF) prevents malicious inputs"),

    ("4. Data Protection",
     "Data masking and anonymization protect sensitive information\n"
     "Differential privacy ensures secure aggregated reporting"),

    ("5. Audit & Monitoring",
     "Comprehensive audit logging tracks all data access\n"
     "SIEM integration enables real-time threat detection"),

    ("6. Infrastructure Security",
     "Virtual Private Cloud (VPC) ensures network isolation\n"
     "Intrusion Detection/Prevention Systems (IDS/IPS) detect attacks"),

    ("7. Backup & Disaster Recovery",
     "Automated encrypted backups\n"
     "Geo-redundant storage ensures business continuity"),
]

for title, description in measures:
    print(f"\n{title}")
    print(f"{description}")
```
