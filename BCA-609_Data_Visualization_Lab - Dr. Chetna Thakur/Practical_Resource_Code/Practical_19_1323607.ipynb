**Program No. 19.  >> Open and Closed Port Visualization**

**Objective:** To visualize basic network security data.

**Tasks:**
- Create a chart showing open and closed ports

**Outcome:** Students will understand visualization of network status.


Since we don’t have a real port-scanner dataset in your folder, we will:
- Create a realistic simulated network scan dataset
- Use common ports (HTTP, HTTPS, SSH, FTP, etc.)
- Mark some as Open and some as Closed
- Visualize them professionally

This mimics real cybersecurity dashboard behavior.

**Step 1:** Import Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
```

**Step 2:** Create Network Port Dataset

```python
# Simulated Port Scan Data
data = {
    "Port": [21, 22, 23, 25, 53, 80, 110, 143, 443, 3306, 8080],
    "Service": [
        "FTP", "SSH", "Telnet", "SMTP", "DNS",
        "HTTP", "POP3", "IMAP", "HTTPS",
        "MySQL", "HTTP-Alt"
    ],
    "Status": [
        "Closed", "Open", "Closed", "Open", "Open",
        "Open", "Closed", "Closed", "Open",
        "Closed", "Open"
    ]
}

df = pd.DataFrame(data)

print(df)
```

**Visualization 1** — Open vs Closed Count (Bar Chart)

```python
status_count = df["Status"].value_counts()

plt.figure(figsize=(7,5))

plt.bar(status_count.index, status_count.values)

plt.title("Open vs Closed Ports",
          fontsize=14,
          fontweight="bold")

plt.xlabel("Port Status")
plt.ylabel("Number of Ports")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Visualization 2** — Port Status by Port Number

```python
plt.figure(figsize=(10,5))

colors = df["Status"].map({"Open": "green", "Closed": "red"})

plt.scatter(df["Port"],
            [1]*len(df),
            c=colors,
            s=200)

for i, txt in enumerate(df["Service"]):
    plt.text(df["Port"][i],
             1.02,
             txt,
             ha='center',
             fontsize=8)

plt.yticks([])
plt.title("Network Port Scan Result",
          fontsize=14,
          fontweight="bold")

plt.xlabel("Port Number")

plt.grid(axis="x", alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt 
import pandas as pd 

ports_data = pd.DataFrame({ 
    'Port': [22, 80, 443, 3306, 5432, 8080, 21, 25, 53, 3389], 
    'Service': ['SSH', 'HTTP', 'HTTPS', 'MySQL', 'PostgreSQL', 'Proxy', 'FTP', 'SMTP', 'DNS', 'RDP'], 
    'Status': ['Open', 'Open', 'Open', 'Closed', 'Open', 'Closed', 'Closed', 'Open', 'Open', 'Closed'] 
    }) 

print("--- Port Scan Results ---") 
print(ports_data.to_string(index=False)) 

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

colors = ['#4CAF50' if s == 'Open' else '#F44336' for s in ports_data['Status']] 
bars = axes[0].barh(ports_data['Service'] + ' (' + ports_data['Port'].astype(str) + ')',
                    [1]*len(ports_data), color=colors, edgecolor='black', linewidth=0.5) 

axes[0].set_title('Port Status — Open vs Closed', fontsize=13, fontweight='bold') 
axes[0].set_xlabel('Status') 
axes[0].set_xticks([]) 

for bar, status in zip(bars, ports_data['Status']): 
    axes[0].text(0.5, bar.get_y()+bar.get_height()/2, status, 
                 ha='center', va='center', fontweight='bold', color='white') 
    
status_counts = ports_data['Status'].value_counts() 
axes[1].pie(status_counts, labels=status_counts.index, autopct='%1.0f%%', colors=['#4CAF50', '#F44336'], startangle=90, textprops={'fontweight': 'bold'}) 
axes[1].set_title('Port Distribution', fontsize=13, fontweight='bold') 

plt.tight_layout() 
plt.show() 

print(f"\nOpen Ports: {status_counts.get('Open',0)} | Closed Ports: {status_counts.get('Closed',0)}") 
print("Security Note: Unused open ports should be closed to reduce attack surface.") 
```
