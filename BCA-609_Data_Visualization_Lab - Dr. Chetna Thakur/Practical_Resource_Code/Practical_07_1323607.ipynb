### Program No. 07.  >> Area Chart for Time-Series Data
**Objective**: To visualize cumulative data over time.

Tasks:
- Create an area chart for yearly profit or rainfall data

**Outcome**: Students will understand growth patterns using area charts.


**Import libraries**

```python
# Tool: Python (Matplotlib)
import pandas as pd
import matplotlib.pyplot as plt
```

**Load Dataset**: `MMDU_Student_Performance`

```python
# Load dataset
df = pd.read_csv("district wise rainfall normal.csv")

# Clean column names (remove extra spaces if any)
df.columns = df.columns.str.strip()

# Group by State and calculate total annual rainfall
state_rainfall = df.groupby("STATE_UT_NAME")["ANNUAL"].sum().sort_values()

# Convert to DataFrame
state_rainfall = state_rainfall.reset_index()
```

**Stage 4: Create Professional Area Chart**

```python
# Set figure size
plt.figure(figsize=(14, 7))

# Create area chart
plt.fill_between(
    state_rainfall["STATE_UT_NAME"],
    state_rainfall["ANNUAL"],
    color="skyblue",
    alpha=0.6
)

# Add line for clarity
plt.plot(
    state_rainfall["STATE_UT_NAME"],
    state_rainfall["ANNUAL"],
    color="blue",
    linewidth=2
)

# Formatting
plt.xticks(rotation=90)
plt.title("Annual Rainfall Distribution by State (Area Chart)", fontsize=16, fontweight='bold')
plt.xlabel("States / UT", fontsize=12)
plt.ylabel("Total Annual Rainfall (mm)", fontsize=12)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

The Area Chart shows:

🔹 Total rainfall contribution of each state

🔹 Relative rainfall distribution

🔹 Cumulative rainfall variation

🔹 States with highest rainfall stand visually dominantHandle Missing Values
