**Step 1:** Import Libraries

```python
import matplotlib.pyplot as plt
import random
import time
```

**Step 2:** Create Live-like Updating Graph

```python
plt.ion()  # Turn on interactive mode

x_values = []
y_values = []

plt.figure(figsize=(8,5))

for i in range(20):  # Simulate 20 live updates
    
    # Simulate new incoming performance value
    new_score = random.randint(60, 95)
    
    x_values.append(i)
    y_values.append(new_score)
    
    plt.clf()
    plt.plot(x_values, y_values, marker='o')
    
    plt.title("Live Academic Performance Monitoring",
              fontsize=14,
              fontweight="bold")
    
    plt.xlabel("Time (Seconds)")
    plt.ylabel("Average Class Score")
    
    plt.ylim(50, 100)
    plt.grid(alpha=0.3)
    
    plt.pause(1)  # 1 second delay

plt.ioff()
plt.show()
```

```python
import matplotlib.pyplot as plt
import random
import time

plt.ion()  # interactive mode

x_values = []
y_values = []

fig, ax = plt.subplots(figsize=(8,5))

line, = ax.plot([], [], marker='o')

ax.set_title("Live Academic Performance Monitoring", fontsize=14, fontweight="bold")
ax.set_xlabel("Time (Seconds)")
ax.set_ylabel("Average Class Score")
ax.set_ylim(50, 100)
ax.set_xlim(0, 20)
ax.grid(alpha=0.3)

for i in range(20):
    
    new_score = random.randint(60, 95)
    
    x_values.append(i)
    y_values.append(new_score)
    
    line.set_data(x_values, y_values)
    
    ax.set_xlim(0, max(20, i+1))
    
    fig.canvas.draw()
    fig.canvas.flush_events()
    
    time.sleep(1)

plt.ioff()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import time

# Load Bitcoin dataset
df = pd.read_csv("bitcoin_2017_to_2023.csv")

# Clean column names
df.columns = df.columns.str.strip()

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    format="%d-%m-%Y %H:%M"
)

# Make sure Date is datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Sort by date
df = df.sort_values("timestamp")

# Use closing price
df["close"] = pd.to_numeric(df["close"], errors="coerce")
df = df.dropna(subset=["close"])

prices = df["close"].values[:50]  # First 50 points for demo
dates = df["timestamp"].values[:50]

plt.ion()

fig, ax = plt.subplots(figsize=(9,6))

line, = ax.plot([], [], marker='o', color='green')
area = None

open_price = prices[0]

ax.set_title("Live Bitcoin Price Monitor", fontsize=14, fontweight="bold")
ax.set_xlabel("Time")
ax.set_ylabel("Price ($)")
ax.grid(alpha=0.3)

ax.axhline(open_price, linestyle="--", alpha=0.5,
           label=f"Open: ${open_price:.2f}")

ax.legend()

x_vals = []
y_vals = []

for i in range(len(prices)):
    
    x_vals.append(i)
    y_vals.append(prices[i])
    
    line.set_data(x_vals, y_vals)
    
    if area:
        area.remove()
        
    area = ax.fill_between(x_vals, y_vals, alpha=0.2)
    
    ax.set_xlim(0, len(prices))
    ax.set_ylim(min(prices)*0.98, max(prices)*1.02)
    
    ax.set_title(f"Live Bitcoin Price Monitor — Update #{i+1}",
                 fontsize=14,
                 fontweight="bold")
    
    fig.canvas.draw()
    fig.canvas.flush_events()
    
    time.sleep(0.2)

plt.ioff()
plt.show()
```
