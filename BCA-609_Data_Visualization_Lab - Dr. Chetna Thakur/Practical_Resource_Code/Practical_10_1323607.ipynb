**Step 2:** Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify
```

**Step 3:** Load and Prepare Dataset

```python
# Load dataset
df = pd.read_csv("student_extended.csv")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Count students per department
dept_counts = df["department"].value_counts()

print(dept_counts)
```

**Step 4:** Create Tree Map

```python
plt.figure(figsize=(10,6))

# Prepare labels
labels = [f"{dept}\n{count} students" 
          for dept, count in zip(dept_counts.index, dept_counts.values)]

# Define sizes
sizes = dept_counts.values

# Color palette
colors = plt.cm.Set3(range(len(sizes)))

# Plot treemap
squarify.plot(
    sizes=sizes,
    label=labels,
    color=colors,
    alpha=0.9
)

plt.title("Department-wise Student Count — Tree Map", fontsize=14, fontweight="bold")
plt.axis("off")

plt.tight_layout()
plt.show()
```

The Tree Map visualization represents department-wise student distribution using proportional rectangular areas.

Each rectangle corresponds to a department, and its size reflects the total number of students enrolled in that department. Larger rectangles indicate higher student enrollment, while smaller rectangles represent comparatively lower enrollment.

This visualization allows quick comparison of categorical data without using axes, making it efficient for hierarchical and proportional data representation.

**Tree maps are especially useful when:**
- Comparing categorical distributions
- Visualizing part-to-whole relationships
- Displaying hierarchical institutional data
- Identifying dominant categories instantly

From the analysis, departments with larger rectangles indicate stronger enrollment presence in the institution.
