**Objective:** To analyze correlation among numerical variables.

- Create a color-coded correlation matrix (heatmap)

Outcome: Students will analyze relationships among multiple variables.


**Step 1:** Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Step 2:** Load and Prepare Dataset

```python
df = pd.read_csv("student_extended.csv")

df.columns = df.columns.str.strip().str.lower()

# Select numerical columns for correlation
numerical_cols = [
    "study_hours_per_day",
    "attendance_percentage",
    "sgpa",
    "percentage",
    "age"
]

data = df[numerical_cols]
```

**Step 3:** Create Correlation Matrix

```python
corr_matrix = data.corr()
print(corr_matrix)
```

**Step 4:** Create Professional Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title("Correlation Heatmap — Student Performance Analysis",
          fontsize=14,
          fontweight="bold")

plt.tight_layout()
plt.show()
```

From this heatmap, we typically observe:
- Study hours show positive correlation with percentage.
- Attendance may moderately influence SGPA.
- Percentage and SGPA usually have strong positive correlation.
- Age generally shows weak correlation with academic performance.
- Variables with near zero correlation are largely independent.

This visualization helps in identifying:
- Key performance drivers
- Strongly related academic factors
- Independent variables
- Multicollinearity risks
