# 📈 Bitcoin Time-Series Professional Analysis
This notebook performs interactive financial analysis on Bitcoin OHLC data.

Includes:
- Close Price Area Chart
- Candlestick Chart
- Volume Analysis
- Moving Average Overlay
- Price + Volume Combined Dashboard


```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
```

```python
# Load dataset
df = pd.read_csv('bitcoin_2017_to_2023.csv')

# Convert timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp')

df.head()
```

```python
import plotly.io as pio
pio.renderers.default = "browser"

fig.show()
```

## 📊 Close Price Area Chart

```python
fig = px.area(df, x='timestamp', y='close', title='Bitcoin Close Price Trend')
fig.update_layout(template='plotly_dark', hovermode='x unified')
fig.show()
```

## 🕯️ Candlestick Chart

```python
fig = go.Figure(data=[go.Candlestick(
    x=df['timestamp'],
    open=df['open'],
    high=df['high'],
    low=df['low'],
    close=df['close']
)])
fig.update_layout(template='plotly_dark', title='Bitcoin Candlestick Chart')
fig.show()
```

## 📊 Volume Analysis

```python
fig = px.bar(df, x='timestamp', y='volume', title='Bitcoin Trading Volume')
fig.update_layout(template='plotly_dark')
fig.show()
```

## 📈 Moving Average (20 Period)

```python
df['MA_20'] = df['close'].rolling(window=20).mean()

fig = go.Figure()
fig.add_trace(go.Scatter(x=df['timestamp'], y=df['close'], mode='lines', name='Close Price'))
fig.add_trace(go.Scatter(x=df['timestamp'], y=df['MA_20'], mode='lines', name='20 Period MA'))
fig.update_layout(template='plotly_dark', hovermode='x unified',
                  title='Bitcoin Price with Moving Average')
fig.show()
```

## 📊 Combined Price & Volume Dashboard

```python
fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                    vertical_spacing=0.05,
                    row_heights=[0.7, 0.3])

fig.add_trace(go.Scatter(x=df['timestamp'], y=df['close'], name='Close Price'), row=1, col=1)
fig.add_trace(go.Bar(x=df['timestamp'], y=df['volume'], name='Volume'), row=2, col=1)

fig.update_layout(template='plotly_dark', height=700,
                  title='Bitcoin Price & Volume Analysis')
fig.show()
```
